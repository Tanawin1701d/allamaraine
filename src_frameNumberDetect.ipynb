```python
import cv2
import os

# Directory containing video files
video_directory = '/media/tanawin/tanawin1701e/project2/chess/vdo'
output_directory = 'out_withframeNumber'

# Ensure the output directory exists
os.makedirs(output_directory, exist_ok=True)

# Loop through each file in the directory
for video_file in os.listdir(video_directory):
    print("prcessing ", video_file)
    # Check if the file is a video file (optional)
    if video_file.endswith('.mp4') or video_file.endswith('.avi'):
        video_path = os.path.join(video_directory, video_file)
        cap = cv2.VideoCapture(video_path)

        if not cap.isOpened():
            print(f"Error opening video file: {video_path}")
            continue

        # Get video details for output file
        fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # Codec for .mp4 format
        fps = cap.get(cv2.CAP_PROP_FPS)  # Frames per second
        frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

        # Create VideoWriter object to save the processed video
        output_path = os.path.join(output_directory, f"processed_{video_file}")
        out = cv2.VideoWriter(output_path, fourcc, fps, (frame_width, frame_height))

        # Process the video
        while cap.isOpened():
            ret, frame = cap.read()
            if not ret:
                break

            # Get the current frame number
            frame_number = int(cap.get(cv2.CAP_PROP_POS_FRAMES))

            # Display the frame number on the frame
            cv2.putText(frame, f'Frame: {frame_number}', (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

            # Write the processed frame to the output video
            out.write(frame)

        # Release the video capture and writer
        cap.release()
        out.release()
        print("prcessing complete", video_file)

print("Processing and saving completed.")
```
