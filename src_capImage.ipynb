```python
import cv2
import os

# Open the video file

for videoFile in os.listdir("vdo"):
    video_path = os.path.join("vdo", videoFile)  # Replace with your video path
    cap = cv2.VideoCapture(video_path)

    # Check if the video was opened successfully
    if not cap.isOpened():
        print("Error: Could not open video.")
    else:
        # Read a single frame
        ret, frame = cap.read()

        if ret:
            # Save the frame as an image
            cv2.imwrite(f'examplePic/{videoFile[0:-4]}.png', frame)
            print("Frame captured and saved as 'captured_frame.jpg'.")
        else:
            print("Error: Could not read a frame.")

    # Release the video capture object
    cap.release()

    # Close any OpenCV windows (if any are open)
    
```
